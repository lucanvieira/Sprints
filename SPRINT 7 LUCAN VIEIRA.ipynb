## Exercici 2 okkkkk

Convertidor de temperatures.
Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re). Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions.


```python
def conversor_de_temperaturas():
       
    tipo = input('¿Qué tipo de temperatura te gustaria convertir? a) Celsius b) Fahrenheit c) Kelvin: ')

    if tipo in ['a', 'b', 'c']:
        try:
            temperatura = input('Dime la temperatura que te gustaría convertir: ')
            temperatura = temperatura.replace(',', '.')  
            temperatura = float(temperatura)

            diccionario_temp = {
                'cf': round((temperatura * 9/5) + 32, ),  
                'ck': round(temperatura + 273.15, 3),  
                'fc': round((temperatura - 32) * 5/9, 3), 
                'fk': round((temperatura - 32) * 5/9 + 273.15, 3),
                'kc': round(temperatura - 273.15, 3), 
                'kf': round((temperatura - 273.15) * 9/5 + 32, 3), 
            }

            if tipo == 'a':
                print(f"{temperatura}° Celsius son: {diccionario_temp['cf']}° Fahrenheit y {diccionario_temp['ck']} Kelvin")
            elif tipo == 'b':
                print(f"{temperatura}° Fahrenheit son: {diccionario_temp['fc']}° Celsius y {diccionario_temp['fk']} Kelvin")
            elif tipo == 'c':
                print(f"{temperatura} Kelvin son: {diccionario_temp['kc']}° Celsius y {diccionario_temp['kf']}° Fahrenheit")
                            

        except ValueError:
            print("Por favor, introduce un número válido para la temperatura.")
    else:
        print("Opción no válida. Por favor selecciona a, b o c.")

conversor_de_temperaturas()
```

## Exercício 3
Contador de palavras de um texto.
Escreva uma função que, dado um texto, mostre o número de vezes que cada palavra aparece.

```python
import re

def remove_pontos(lista):
    return re.sub(r'[^\w\s]', '', lista.lower())

def contador_de_palabras():

    formato_de_texto = input('Cual es el formato de tu texto? [a] - Quiero Digitarlo   [b] formato (.txt): ')

    if formato_de_texto in ['a', 'b']:
        try:
            if formato_de_texto == 'a':
                texto = input('Digite tu texto: ')
                print(f'El texto digitado es: {texto}')
            elif formato_de_texto == 'b':
                camino = input('Digite el camino de el archivo: ')
                with open(camino, 'r', encoding='utf-8') as contenido:
                    texto = contenido.read()
    
        except Exception as e:
            print(f'Ocurrió un error: {e}')
    else:
        print("Opción no válida.")

    lista = texto.split() 
   
    texto_ordenado = []
    
    for linhas in lista:
            palavra = remove_pontos(linhas).strip().lower().split()
            texto_ordenado.extend(palavra)

   
    contagem_do_texto = {}

    for palavras in texto_ordenado:
        if palavras  in contagem_do_texto:
            contagem_do_texto[palavras] += 1
        else:
            contagem_do_texto[palavras] = 1
    

    print('\nLa contagem de palabras és esta:')
    return contagem_do_texto

contador_de_palabras()
```

```python
reverse_dictionary = {'x': 'apple', 'y': 'banana', 'z': 'banana'}

def reves_dic_error():
    
    print(reverse_dictionary)
    
    diccionario_inv = {}
    for key, value in reverse_dictionary.items():
        if value in diccionario_inv:
            print(f"\n¡¡¡ERROR¡¡¡ TENEIS EL VALOR: '{value}' DUPLICADO.")
        else:
            diccionario_inv[value] = key
    
    print("\nEL DICCIONARIO INVERTIDO ES:")

    return diccionario_inv

reves_dic_error()
```

Exemple de la llista que rep el client: [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]


**Exercício 2**  
**Conversão de tipos de dados**

O cliente recebe uma lista de dados e precisa gerar duas listas: a primeira onde estarão todos os elementos que puderam ser convertidos em floats e a outra onde estarão os elementos que não puderam ser convertidos. Exemplo da lista que o cliente recebe: [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]



```python
lista_floats = ['1.3', 'one', '1e10', 'seven', '3-1/2', ('2', 1, 1.4, 'not-a-number'), [1, 2, '3', '3.4']]

def conversor_float_no_float():
    float_ = []
    n_float = []

    def conversor_float(value):
        try:
            return float(value)
        except (ValueError, TypeError):
            return None

    for item in lista_floats:
        if isinstance(item, (list, tuple)):
            for subitem in item:
                convert_value = conversor_float(subitem)
                if convert_value is not None:
                    float_.append(convert_value)
                else:
                    n_float.append(subitem)
        else:
            convert_value = conversor_float(item)
            if convert_value is not None:
                float_.append(convert_value)
            else:
                n_float.append(item)

    return {"floats": float_, "nofloats" : n_float}

conversor_float_no_float()
```
